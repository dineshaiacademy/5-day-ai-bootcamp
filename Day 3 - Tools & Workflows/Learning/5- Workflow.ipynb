---

## The simplest definition

> **A workflow is a predefined sequence of steps arranged to transform an input into an expected output.**

A workflow can contain:

- Python functions
- APIs
- databases
- LLM calls
- AI tools
- validation
- conditions
- human approval

### The key idea

```text
INPUT
  ↓
STEP 1
  ↓
STEP 2
  ↓
STEP 3
  ↓
OUTPUT
```

The important word is **predefined**.

In a workflow, the developer decides the process ahead of time.

---

## Tool vs Workflow

**Tool = one capability**

```text
calculate()
search_web()
send_email()
```

**Workflow = multiple steps connected together**

```text
Get request
   ↓
Search information
   ↓
Summarize information
   ↓
Create email
   ↓
Ask for approval
   ↓
Send email
```

> **A tool performs an action. A workflow defines how multiple actions are performed together.**



# 1. Start with a real-world example

Imagine a student submits:

> "Please prepare a summary of today's customer complaints and send it to my manager."

An LLM alone can write a summary **if the complaints are already in its context**.

But the complete task requires several capabilities:

```text
1. Get customer complaints
2. Filter today's complaints
3. Summarize them
4. Create an email
5. Get approval
6. Send the email
```

That sequence is a **workflow**.

### Workflow representation

```text
Customer Complaints
        ↓
   Get Data
        ↓
  Filter Today
        ↓
   LLM Summary
        ↓
  Draft Email
        ↓
 Human Approval
        ↓
   Send Email
```

Notice something important:

**The workflow does not need an agent to exist.**

The developer can define this sequence explicitly.



# 2. Workflow vs Tool vs Agent

This distinction removes a lot of confusion.

| Concept | What it means | Who defines the next action? |
|---|---|---|
| **Tool** | A capability/action | Application calls it |
| **Workflow** | Predefined sequence of actions | Developer |
| **Agent** | System where the model can decide actions toward a goal | Model + application |
| **LLM** | Model that generates/predicts outputs | Model |

### Example

**Tool:**

```python
calculate(20, 5, "multiply")
```

**Workflow:**

```text
Get price
   ↓
Calculate discount
   ↓
Format result
```

**Agent:**

```text
Goal: "Find the best laptop under ₹70,000."

The system may decide:
   ↓
Search products
   ↓
Compare results
   ↓
Search again if information is missing
   ↓
Calculate prices
   ↓
Answer
```

An agent can use workflows and tools, but **workflow and agent are not the same thing**.



# 3. Build our first workflow — without AI

Before adding an LLM, let's understand a workflow using ordinary Python.

Our business process:

```text
Number
  ↓
Square the number
  ↓
Add 10
  ↓
Format the result
```

This is a workflow because the sequence is explicitly defined.


````python

def square(number):
    return number * number


def add_ten(number):
    return number + 10


def format_result(number):
    return f"The final result is {number}"


def simple_workflow(number):
    step1 = square(number)
    step2 = add_ten(step1)
    step3 = format_result(step2)
    return step3


print(simple_workflow(5))
````


## What happened?

For input `5`:

```text
Input: 5
  ↓
square(5)
  ↓
25
  ↓
add_ten(25)
  ↓
35
  ↓
format_result(35)
  ↓
"The final result is 35"
```

The developer controls the sequence.

There is no ambiguity about what happens next:

```text
Step 1 → Step 2 → Step 3
```

That is a **deterministic workflow**.



# 4. Sequential workflow

The most basic workflow pattern is **sequential execution**.

```text
A → B → C → D
```

Example:

```text
Read document
    ↓
Extract text
    ↓
Summarize
    ↓
Save summary
```

Each step starts after the previous step produces its output.


````python

def read_document():
    return "Customer reported a payment issue. Support resolved it in 20 minutes."


def summarize(text):
    return "Payment issue reported and resolved in 20 minutes."


def save_summary(summary):
    return f"Saved summary: {summary}"


def sequential_workflow():
    document = read_document()
    summary = summarize(document)
    result = save_summary(summary)
    return result


print(sequential_workflow())
````


# 5. A workflow can contain an LLM

A common misconception is:

> "If a workflow contains an LLM, it becomes an agent."

**No.**

An LLM can simply be one step inside a predefined workflow.

For example:

```text
Read document
      ↓
LLM summarizes document
      ↓
Save summary
```

The developer still decides the sequence.

The LLM is being used as a component.


````python

# This function represents an LLM step.
# In a real application, it would call an LLM API.

def llm_summarize(text):
    return "Summary: " + text[:60] + "..."


def document_workflow():
    document = read_document()
    summary = llm_summarize(document)
    result = save_summary(summary)
    return result


print(document_workflow())
````


# 6. Tools inside a workflow

Now connect the previous lesson about tools.

Suppose we have these tools:

```text
get_customer_data()
calculate_discount()
send_email()
```

A workflow can call them in a fixed sequence:

```text
get_customer_data
       ↓
calculate_discount
       ↓
send_email
```

The tools are **capabilities**.

The workflow defines **when and in what order those capabilities are used**.


````python

def get_customer_data():
    return {
        "customer": "Rahul",
        "purchase_amount": 5000
    }


def calculate_discount(amount):
    return amount * 0.90


def send_email(customer, amount):
    return f"Email prepared for {customer}: discounted amount = ₹{amount:.2f}"


def customer_discount_workflow():
    customer_data = get_customer_data()

    discounted_amount = calculate_discount(
        customer_data["purchase_amount"]
    )

    email_result = send_email(
        customer_data["customer"],
        discounted_amount
    )

    return email_result


print(customer_discount_workflow())
````


# 7. Conditional workflow

Real workflows are not always straight lines.

Sometimes:

```text
IF condition is true
    ↓
Step A
ELSE
    ↓
Step B
```

Example:

```text
Read order
    ↓
Is order > ₹10,000?
    ├── YES → Manager approval
    └── NO  → Process automatically
```

The important point:

**The condition is predefined by the developer.**


````python

def process_order(amount):
    print(f"Order amount: ₹{amount}")

    if amount > 10000:
        return "Manager approval required"
    else:
        return "Order can be processed automatically"


print(process_order(7500))
print(process_order(15000))
````


# 8. Parallel workflow

Sometimes independent tasks can run at the same time.

Example:

```text
             ┌→ Get weather ─────┐
User request ┤                   ├→ Combine results
             └→ Get news ────────┘
```

Weather and news do not depend on each other.

So the workflow can execute them in parallel.

### Sequential

```text
A → B → C
```

### Parallel

```text
     ┌→ A ─┐
Start     ├→ C
     └→ B ─┘
```

Parallel execution is mainly an **engineering optimization** when steps are independent.



# 9. Human-in-the-loop workflow

Some actions should require human approval.

Example: sending an important email.

```text
Generate email
      ↓
Human reviews
      ↓
Approved?
   ├── YES → Send email
   └── NO  → Stop / Edit
```

This is still a workflow.

The human is simply one step in the process.



# 10. The most important distinction — Workflow vs Agent

Let's compare the same business goal.

## Workflow approach

Goal:

> "Prepare and send a customer report."

Developer defines:

```text
1. Get customer data
2. Filter today's data
3. Summarize
4. Create report
5. Ask for approval
6. Send report
```

The sequence is predefined.

```text
A → B → C → D → E → F
```

---

## Agent approach

Goal:

> "Prepare a useful customer report."

The application gives the model tools such as:

```text
get_customer_data
search_database
calculate_metrics
create_report
send_email
```

The model can determine which action to request next based on the current state.

Conceptually:

```text
Goal
 ↓
Model decides action
 ↓
Tool
 ↓
Observe result
 ↓
Model decides next action
 ↓
Tool
 ↓
...
```

### Therefore

> **Workflow = developer-defined process.**

> **Agent = model-driven decision-making process toward a goal.**



# 11. A very clear analogy

Think about a restaurant.

### Recipe = Workflow

A chef follows:

```text
1. Cut vegetables
2. Heat pan
3. Add oil
4. Add vegetables
5. Cook for 5 minutes
6. Serve
```

The sequence is predefined.

### Kitchen equipment = Tools

```text
Knife
Pan
Oven
Mixer
```

Each provides a capability.

### Chef deciding what to do = Agent-like behavior

The chef has a goal:

> "Prepare dinner for four people."

Depending on ingredients and the situation, the chef can decide what action to take next.

### AI engineering mapping

```text
Tools    → Capabilities
Workflow → Predefined process
Agent    → Goal + model-driven decisions
```

# 13. Workflow patterns students should know

| Pattern | Structure | Example |
|---|---|---|
| Sequential | A → B → C | Read → summarize → save |
| Conditional | IF → A / B | High value → approval |
| Parallel | A + B → C | Search web + database → combine |
| Loop | A → repeat → stop | Process every document |
| Human-in-loop | A → human → B | Draft → approve → send |
| Retry | A → failure → retry | API call → retry |
| Branching | A → choose path | Request type → specialized process |

These are **workflow patterns**.

Frameworks such as LangGraph can help implement complex workflows, but the concept exists independently of any framework.



# 14. Mini exercise — identify the workflow

### Scenario

A company receives an invoice PDF.

The system must:

```text
1. Read the PDF
2. Extract invoice details
3. Check the vendor in the database
4. Calculate tax
5. If amount > ₹1,00,000, ask manager approval
6. Otherwise save the invoice
7. Send confirmation
```

### Questions

**Q1. What are the tools?**

Possible answers:

```text
read_pdf()
extract_invoice()
lookup_vendor()
calculate_tax()
save_invoice()
send_email()
```

**Q2. What is the workflow?**

```text
Read
 ↓
Extract
 ↓
Lookup
 ↓
Calculate
 ↓
IF amount > threshold
 ├── YES → Manager approval
 └── NO  → Save
 ↓
Confirmation
```

**Q3. Is this automatically an agent?**

No.

If the sequence and decision rules are explicitly predefined, it is a workflow.



# 15. Final mental model

Remember this hierarchy:

```text
                 AI APPLICATION
                       │
          ┌────────────┴────────────┐
          │                         │
       WORKFLOW                   AGENT
          │                         │
    predefined flow         model-driven decisions
          │                         │
      ┌───┴───┐                 ┌───┴───┐
      │       │                 │       │
     LLM    TOOLS              LLM    TOOLS
```

And the simplest definitions:

### 🔧 Tool
> **Something the application can execute to perform a capability.**

### 🔄 Workflow
> **A predefined sequence of steps that transforms an input into an output.**

### 🤖 Agent
> **A system where an AI model can decide which actions/tools to take next to achieve a goal.**
