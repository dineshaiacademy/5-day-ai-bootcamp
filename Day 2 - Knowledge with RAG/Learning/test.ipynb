## 🔐 Managing Secrets (VS Code + Colab)

This cell loads API keys the same way whether the notebook runs locally in VS Code or in Google Colab.

- **VS Code / local**: uses the `.env` file at the repo root (via `.env.example` as a template). It is gitignored, so it never gets committed.
- **Colab**: click the 🔑 key icon in the left sidebar, add your keys there, and enable notebook access.

```python
%pip install -q python-dotenv
```

```python
import os

def get_secret(key_name):
    try:
        from google.colab import userdata
        return userdata.get(key_name)
    except ImportError:
        from dotenv import load_dotenv, find_dotenv
        load_dotenv(find_dotenv())
        return os.getenv(key_name)

OPENAI_API_KEY = get_secret("OPENAI_API_KEY")
GROQ_API_KEY = get_secret("GROQ_API_KEY")
```
