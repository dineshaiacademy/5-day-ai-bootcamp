The subtask requires defining a Python function named `calculator` with specified arguments, arithmetic operations, and error handling for division by zero. This code block will implement the function as described.



```python
def calculator(num1, num2, operation):
    """
    Performs basic arithmetic operations between two numbers.

    Args:
        num1 (int or float): The first number.
        num2 (int or float): The second number.
        operation (str): The arithmetic operation ('add', 'subtract', 'multiply', 'divide').

    Returns:
        int or float: The result of the operation.
        str: An error message if division by zero occurs or an invalid operation is provided.
    """
    if operation == 'add':
        return num1 + num2
    elif operation == 'subtract':
        return num1 - num2
    elif operation == 'multiply':
        return num1 * num2
    elif operation == 'divide':
        if num2 == 0:
            return "Error: Division by zero is not allowed."
        return num1 / num2
    else:
        return "Error: Invalid operation. Please choose 'add', 'subtract', 'multiply', or 'divide'."

print("The 'calculator' function has been defined.")
```

## Implement Agent's Tool Usage Logic

### Subtask:
Add logic within the agent class to decide when and how to use its available tools based on an input query or task. This will involve parsing the input and matching it to a tool's capabilities, then executing the tool and processing its output.


**Reasoning**:
Refactor the Agent class to implement rule-based query parsing, parameter extraction, dynamic tool execution using the metadata dictionary, and output formatting as requested.



```python
import re

class Agent:
    def __init__(self, tools):
        self.tools = {tool['name']: tool for tool in tools}
        print(f"Agent initialized with tools: {list(self.tools.keys())}")

    def handle_query(self, query):
        print(f"\nProcessing Query: '{query}'")

        # Simple regex pattern to capture common arithmetic query patterns
        # e.g., "add 5 and 3", "divide 10 by 2", "multiply 4 by 5.5"
        pattern = r'(add|subtract|multiply|divide)\s+([-+]?\d*\.?\d+)\s+(?:and|by)\s+([-+]?\d*\.?\d+)'
        match = re.search(pattern, query.lower())

        if match:
            operation = match.group(1)
            num1 = float(match.group(2))
            num2 = float(match.group(3))

            # Convert to int if it's a whole number for cleaner output
            if num1.is_integer(): num1 = int(num1)
            if num2.is_integer(): num2 = int(num2)

            print(f"Parsed command: operation='{operation}', num1={num1}, num2={num2}")

            # Check if we have a tool matching 'calculator'
            if 'calculator' in self.tools:
                calculator_tool = self.tools['calculator']
                # Dynamic invocation of the stored function reference
                result = calculator_tool['function'](num1, num2, operation)
                return f"[Tool Used: calculator] Result of {operation}ing {num1} and {num2} is: {result}"
            else:
                return "Error: Calculator tool is not available to this agent."
        else:
            return "I'm sorry, I could not parse a valid arithmetic operation from your query."

# Initialize the agent with our calculator tool metadata
agent = Agent([calculator_tool_metadata])

# Test cases to demonstrate tool usage
queries = [
    "Can you add 15 and 27?",
    "Please divide 100 by 4",
    "What is multiply 3.5 by 2?",
    "Can you divide 5 by 0?",
    "What is the weather today?"
]

for q in queries:
    response = agent.handle_query(q)
    print(f"Agent Response: {response}")
```

### 1. Define the Web Search Tool & Metadata
We will define a Python function `web_search` that simulates an API call or web search, and we will package it with MCP-style metadata.

```python
def web_search(query):
    """
    Simulates a web search or an external API request for informational queries.
    """
    query_lower = query.lower()
    if "weather" in query_lower:
      return "Sunny, 22°C with a light breeze."
    elif "distance to the moon" in query_lower or "moon" in query_lower:
      return "The average distance to the Moon is approximately 384,400 kilometers (238,855 miles)."
    elif "capital of france" in query_lower:
      return "The capital of France is Paris."
    else:
      return f"Search results for '{query}': No highly relevant pages found. Try adjusting your query."

web_search_metadata = {
    'name': 'web_search',
    'description': 'A tool that performs informational web searches and external API queries for weather, geography, and general knowledge questions.',
    'parameters': {
        'query': {
            'type': 'string',
            'description': 'The query or topic to search on the web.',
            'required': True
        }
    },
    'function': web_search
}

print("Web search tool and metadata defined successfully.")
```

### 2. Update the Agent Class to Support Multi-Tool Dispatching
We will now refactor the `Agent` class to inspect the input query, identify the correct tool, unpack the arguments dynamically based on metadata schemas, and run the matching tool.

```python
import re

class MultiToolAgent:
    def __init__(self, tools):
        # Store tools mapped by their name
        self.tools = {tool['name']: tool for tool in tools}
        print(f"Agent initialized with tools: {list(self.tools.keys())}")

    def handle_query(self, query):
        print(f"\nProcessing Query: '{query}'")

        # 1. Routing Logic: Determine if it's a calculator or search task
        math_pattern = r'(add|subtract|multiply|divide)\s+([-+]?\d*\.?\d+)\s+(?:and|by|from)\s+([-+]?\d*\.?\d+)'
        math_match = re.search(math_pattern, query.lower())

        if math_match and 'calculator' in self.tools:
            operation = math_match.group(1)
            num1 = float(math_match.group(2))
            num2 = float(math_match.group(3))

            if num1.is_integer(): num1 = int(num1)
            if num2.is_integer(): num2 = int(num2)

            # Execute using metadata reference
            tool = self.tools['calculator']
            result = tool['function'](num1, num2, operation)
            return f"[Tool Used: {tool['name']}] {result}"

        # 2. If it's an informational query, fall back to web_search tool if available
        elif 'web_search' in self.tools:
            tool = self.tools['web_search']
            result = tool['function'](query)
            return f"[Tool Used: {tool['name']}] {result}"

        else:
            return "I am sorry, I do not have a suitable tool registered to resolve this query."
```

### 3. Initialize and Test the Extended Agent
Let's test both tool routes with the updated agent using the test suite.

```python
# Instantiate MultiToolAgent with both our calculator and web_search metadata
advanced_agent = MultiToolAgent([calculator_tool_metadata, web_search_metadata])

test_queries = [
    "What is the distance to the moon?",
    "Can you add 15 and 27?",
    "How is the weather today?",
    "Please divide 100 by 4",
    "What is the capital of France?"
]

for q in test_queries:
    response = advanced_agent.handle_query(q)
    print(f"Agent Response: {response}")
```
